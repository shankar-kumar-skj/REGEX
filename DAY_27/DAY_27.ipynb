### REMOVE HTML TAGS

```python
import re

def remove_html_tags(text):
    pattern=re.compile('<.*?>')
    return pattern.sub(r' ',text)
```

```python
text="<html><body><p>Movies1</p><p>Click here to <a href='https://google.com'>download</a></p>"
```

```python
remove_html_tags(text)
```

```python
# now we remove tags from IMDB Datasets
df['review']=df['review'].apply(remove_html_tags)
```

### IF WE WANT TO REMOVE URLS FROM TEXT THEN

```python
def remove_url(text):
    pattern=re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r' ',text)
```

```python
text4='for notebook click https://www.google.com/skj'
```

```python
remove_url(text4)
```

### STEP-4 REMOVE PUNCTUATION

```python
import string 
string.punctuation
```

```python
exclude=string.punctuation
```

```python
def remove_punctuation(text):
    for char in exclude:
        text=text.replace(char,' ')
    return text
```

```python
text1="Q3 WHAT IS PUNCTUATION"
```

```python
remove_punctuation(text1)
```

```python
df['review']=df['review'].apply(remove_punctuation)
```

#### STOPWORDS

##### MANUALLY WORK

```python
def remove_stopwords(text):
    new_text=[]
    for word in text.split():
        if word in set(stopwords.words('english')):
            new_text.append('')
        else:
            new_text.append(word)

    x=new_text[:]
    new_text.clear()
    return "".join(x)

remove_stopwords('probability my all-time favourite movies, a story of selflessness')
```

```python
# TF(Term Frequencey) 
# 2. IDF(Inverse Document Frequencey)
```

```python
# TF(t,d) = (Number of occurences of term t in document d)/(Total number of terms in the document d) 
# example ==>
# d1 = people watch cricket . 

# tf(people,d1) = 1/3 

# d2 = match was match. 
# tf(match , d2) = 2/3 
```

```python
# IDF(t) = loge(Total number of documents in the corpus)/(Number of documents with term t in them) 
# here corpus means number of sentences

# Example ==>
# d1 = people watch cricket . 
# d2 = cricket was cricket . 
# d3 = people write comment .
# d4 = cricket write comment . 

# idf(cricket) = log(4/4)
# idf(people) = log(4/2) 
```

```python
# we can use this by apply feature_extraction method in sklearn 
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf=TfidfVectorizer()
tfidf.fit_transform(df['text']).toarray()
```

```python
print(tfidf.idf_)
print(tfidf.get_feature_names_out())
```
